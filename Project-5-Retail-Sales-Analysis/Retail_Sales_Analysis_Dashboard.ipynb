```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
sb.barplot(x="Category",
           y="Sales",
           data=df)

mp.title("Sales by Category")

mp.show()
```

```python
line_data = df.groupby("Region")["Sales"].sum()

mp.plot(line_data.index,
        line_data.values,
        marker="o")

mp.title("Region-wise Sales")

mp.xlabel("Region")

mp.ylabel("Sales")

mp.show()
```

```python
sb.histplot(df["Sales"], kde=True)

mp.title("Sales Distribution")

mp.show()
```

```python
sb.scatterplot(x="Sales",
               y="Profit",
               data=df)

mp.title("Sales vs Profit")

mp.show()
```
