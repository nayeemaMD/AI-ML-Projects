```python
sb.boxplot(x=df["total_sales_price"])

mp.title("Boxplot of Total Sales Price")
mp.show()
```

```python
Q1 = df["total_sales_price"].quantile(0.25)
Q3 = df["total_sales_price"].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
```

```python
IQR = Q3 - Q1

print("IQR:", IQR)
```

```python
lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[(df["total_sales_price"] < lower_limit) | (df["total_sales_price"] > upper_limit)]

outliers.shape
```

```python
df = df[(df["total_sales_price"] >= lower_limit) & (df["total_sales_price"] <= upper_limit)]
```

```python
sb.boxplot(x=df["total_sales_price"])

mp.title("Boxplot After Removing Outliers")

mp.show()
```
