```python
sb.boxplot(x=df["alcohol"])

mp.title("Boxplot of Alcohol")

mp.show()
```

```python
Q1 = df["alcohol"].quantile(0.25)

Q3 = df["alcohol"].quantile(0.75)

print(Q1)

print(Q3)
```

```python
IQR = Q3 - Q1

print(IQR)
```

```python
lower_limit = Q1 - (1.5 * IQR)

upper_limit = Q3 + (1.5 * IQR)

print(lower_limit)

print(upper_limit)
```

```python
outliers = df[(df["alcohol"] < lower_limit) |
              (df["alcohol"] > upper_limit)]

outliers.shape
```

```python
df = df[(df["alcohol"] >= lower_limit) &
        (df["alcohol"] <= upper_limit)]
```

```python
sb.boxplot(x=df["alcohol"])

mp.title("Alcohol After Removing Outliers")

mp.show()
```

```python
sb.barplot(x="quality",
           y="alcohol",
           data=df)

mp.title("Average Alcohol by Wine Quality")

mp.show()
```

```python
line_data = df.groupby("quality")["alcohol"].mean()

mp.plot(line_data.index,
        line_data.values,
        marker="o")

mp.title("Wine Quality vs Average Alcohol")

mp.xlabel("Quality")

mp.ylabel("Average Alcohol")

mp.show()
```
