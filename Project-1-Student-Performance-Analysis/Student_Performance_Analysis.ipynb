```python
sb.boxplot(x=df["Exam_Score"])

mp.title("Boxplot of Exam Score")

mp.show()
```

```python
Q1 = df["Exam_Score"].quantile(0.25)

Q3 = df["Exam_Score"].quantile(0.75)

print("Q1 =", Q1)
print("Q3 =", Q3)
```

```python
IQR = Q3 - Q1

print("IQR =", IQR)
```

```python
lower_limit = Q1 - (1.5 * IQR)

upper_limit = Q3 + (1.5 * IQR)

print("Lower Limit =", lower_limit)

print("Upper Limit =", upper_limit)
```

```python
outliers = df[(df["Exam_Score"] < lower_limit) |
              (df["Exam_Score"] > upper_limit)]

outliers.shape
```

```python
df = df[(df["Exam_Score"] >= lower_limit) &
        (df["Exam_Score"] <= upper_limit)]
```

```python
sb.boxplot(x=df["Exam_Score"])

mp.title("Boxplot After Removing Outliers")

mp.show()
```

```python
corr = df.select_dtypes(include=np.number).corr()

corr
```

```python
mp.figure(figsize=(10,8))

sb.heatmap(corr,
           annot=True,
           cmap="coolwarm")

mp.title("Correlation Heatmap")

mp.show()
```

```python
sb.barplot(x="Gender", y="Exam_Score", data=df)

mp.title("Average Exam Score by Gender")

mp.xlabel("Gender")

mp.ylabel("Exam Score")

mp.show()
```

```python
line_data = df.groupby("Hours_Studied")["Exam_Score"].mean()

mp.plot(line_data.index,
        line_data.values,
        marker="o")

mp.title("Hours Studied vs Average Exam Score")

mp.xlabel("Hours Studied")

mp.ylabel("Average Exam Score")

mp.show()
```

```python
sb.histplot(df["Exam_Score"], kde=True)

mp.title("Distribution of Exam Scores")

mp.xlabel("Exam Score")

mp.ylabel("Frequency")

mp.show()
```

```python
sb.scatterplot(x="Hours_Studied",
               y="Exam_Score",
               data=df)

mp.title("Hours Studied vs Exam Score")

mp.xlabel("Hours Studied")

mp.ylabel("Exam Score")

mp.show()
```
